## Question 1: Data Preparation and Cleaning

```python
# Q1 — Load dataset and basic cleaning
import pandas as pd

# Load the raw dataset (replace 'your_dataset.csv' with your actual file path)
df = pd.read_csv('C:\Users\DELL\Desktop\ING 3rd year\securite\S7\MLDLS\lab2\network_data.csv')

# Convert specific columns to numeric
numeric_cols = ['src_port', 'dst_port', 'packets', 'bytes', 'cpu_percent', 'mem_percent', 'anomaly_score']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Convert datetime-related columns
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
df['last_seen'] = pd.to_datetime(df['last_seen'], errors='coerce')

# Normalize label column to categorical type
df['label'] = df['label'].astype('category')

# Display summary
print(" Data loaded and cleaned successfully")
print(df.info())
df.head()
```
